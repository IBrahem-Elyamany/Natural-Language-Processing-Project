```python
# Load the same Embedding Model as DB vector

embedding_model = SentenceTransformer('all-MiniLM-L6-v2', device='cpu')
```

```python
source_db = "/kaggle/input/datasets/roaahussein/db-vector/my_vector_db"
destination_db = "/kaggle/working/my_vector_db"

if not os.path.exists(destination_db):
    shutil.copytree(source_db, destination_db)
    print("Database copied to working directory successfully!")
else:
    print("Database already exists in working directory.")

db_path = destination_db

# Connect to the existing Vector DB
chroma_client = chromadb.PersistentClient(path=db_path)
collection = chroma_client.get_collection(name="cv_matching")
print(f"Connected! Total CVs: {collection.count()}")
```

```python
# ---- Ollama Helper ----
def call_ollama(prompt, model="mistral"):
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={"model": model, "prompt": prompt, "stream": False}
    )
    return response.json()["response"]
```

```python
# ---- Main Function ----
def get_job_recommendations(user_input):
    # Step A: Intelligent Extraction of JD and N
    extraction_prompt = f"""
    Analyze this user request: "{user_input}"
    
    Extract:
    1. The core Job Description (JD).
    2. The number of candidates requested (N). If not specified, default to 3.

    Respond ONLY in this exact format with nothing else:
    JD: [the job description text] | N: [number]
    """
    extraction_res = call_ollama(extraction_prompt)

    try:
        jd_text = extraction_res.split("|")[0].replace("JD:", "").strip()
        top_n   = int(extraction_res.split("|")[1].replace("N:", "").strip())
    except:
        jd_text = user_input
        top_n   = 3

    # Step B: Embedding the extracted Job Description
    # - Using the same model to ensure vector compatibility
    query_vector = embedding_model.encode([jd_text]).tolist()

    # Step C: Retrieval from ChromaDB
    results = collection.query(
        query_embeddings=query_vector,
        n_results=top_n
    )

    # Step D: Context Preparation
    context_docs = "\n---\n".join(results['documents'][0])

    # Step E: Final Generation
    final_prompt = f"""
    You are a professional HR Recruiter. Your task is to evaluate candidates for the following job.
    JOB DESCRIPTION:{jd_text}
    NUMBER OF POSITIONS: {top_n}
    CANDIDATE CVs FROM DATABASE:{context_docs}
    
    INSTRUCTIONS:
    - Select the top {top_n} candidates
    - Ranks all {top_n} candidates from most to least suitable
    - Use the candidate's REAL NAME from their CV (not A, B, or numbers)
    - For each candidate provide:
    1. Full Name
    2. For each candidate explains: why they fit, their key strengths, and any gaps 
    3. Top 3 matching skills
    4. End with a final hiring recommendation

    OUTPUT FORMAT (follow this exactly):
    Rank 1: [Full Name]
    - Why they fit: ...
    - Key strengths: ...
    - Gaps: ...
    - Top skills: skill1, skill2, skill3

    Rank 2: [Full Name]
    - Why they fit: ...
    - Key strengths: ...
    - Gaps: ...
    - Top skills: skill1, skill2, skill3

 
    My Personal Recommendation:
    As an HR professional, I strongly recommend [name] because [specific reason based on the analysis above].
    """
    report = call_ollama(final_prompt)

    return jd_text, top_n, report
```

```python
# --- Example Execution ---
# user_query = "I need 4 software engineers with experience in React and Node.js"
# user_query = "Seeking a 5 Senior Consultant to provide expert advice and strategic solutions. Must have experience in business analysis and project management."
# user_query = "Seeking a Senior Consultant to provide expert advice and strategic solutions. Must have experience in business analysis and project management."
# user_query = "i need 10 candidates suitable for machine learning engineering"
# user_query = "can you recommend 4 candidates from your CVs database has experience in data engineering ?"
user_query = "i am searching for a professional software engineer with 3 years experience and problem solving skill , provide 4 suggestions"
extracted_jd, n, report = get_job_recommendations(user_query)

print(f"Extracted JD: {extracted_jd}")
print(f"Top N requested: {n}")
print("\n--- Final Recruitment Report ---\n")
print(report)
```
